```python
from sklearn import datasets 
iris = datasets.load_iris()
```

```python
x= iris['data']
y=iris['target']
```

```python
from sklearn.linear_model import LogisticRegression 
```

```python
logit = LogisticRegression(max_iter= 1000)
```

```python
print(logit.fit(x,y))
print(logit.score(x,y))
```
