```python
import numpy as np
import pandas as pd
from scipy import stats


# this is an array for example
speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

print(speed)
x= np.mean(speed)
y=np.median(speed)
z=stats.mode(speed)


print("mean of the speeds is = ",format(x,".4f"))
print('Median of the speeds is = ', format(y,'.4f'))
print('mode of the speeds is = ', format(z.mode,'.4f'))
```
