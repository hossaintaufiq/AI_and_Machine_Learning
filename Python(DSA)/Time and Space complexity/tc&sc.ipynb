Check Palindrome

```python
n = int(input("Enter a number: "))

num= n; 
def isPalindrome(num): 
  result=0; 
  
  while num>0: 
    last_digit= num%10
    result= result*10+last_digit
    num=num//10 

  return n==result


print(isPalindrome(n))  
```

Armstron number

```python
from math import *
n= int(input("Enter a number: "))

num= n;
def isArmstrong(num):
    result=0; 
    number_count= floor(log10(num)+1)
    while num>0: 
        last_digit= num%10 
        result= result+ last_digit**number_count
        num=num//10 
    return n==result

print(isArmstrong(n)) 
```

All the Factors of a Given Number -> traditional way Time complexity = O(n)

```python
n= int(input("Enter a number: "))

def AllFactors(n): 
    print('factors of',n, 'are: ', end=' ')
    for i in range(1,n+1):
        if n%i==0:
            print(i,end=' ')

AllFactors(n)
```

Better solution -> Time complexity O(n/2) ->O(n) , Space complexity->O(k)

```python
n= int(input("Enter a number: "))

def AllFactors(n):
    result= []
    for i in range(1,n//2 +1):
        if n%i==0:
            result.append(i)
    result.append(n)
    return result
print(AllFactors(n))
```

Optimal Solution

```python
from math import *
from unittest import result
n= int(input("Enter a number: "))

def AllFactors(n):
    result= []
    for i in range(1, int(sqrt(n))+1):
        if n%i==0: 
            result.append(i)
            if i != n // i:
                result.append(n//i)
    result.sort()
    return result

AllFactors(n)
```

    Frequency map/dictionary 
Store frequency in a Dictionary

time complexity = O(n)

```python
nums=[5,6,7,7,1,9 ,111,1,1,5,1,1]

freq_map= {}         #we can use dictionary to store the frequency of each number in the list = dict()

for i in range(0, len(nums)):
    if nums[i] in freq_map:
        freq_map[nums[i]]+=1
    else: 
        freq_map[nums[i]]=1


print(max(freq_map, key=freq_map.get))  #this will return the key with the maximum value in the dictionary
```

using Hashmap 

```python
nums=[5,6,7,7,1,9 ,111,1,1,5,1,1]

hash_map= {}         #we can use dictionary to store the frequency of each number in the list = dict()
n = len(nums)

for i in range(0, n):
    hash_map[nums[i]] = hash_map.get(nums[i],0)+1
```

```python
#Brute force 
#Time complexity: O(n*m) where n is the length of the first list and m is the length of the second list or O(n^2) if both lists are of equal length

n= [5,3,2,2,1,5,5,7,5,10]
m= [10,111,1,9,5,67,2]

for num in m: 
    count =0 
    for x in n: 
        if x==num : 
            count+=1
    print(count, end=' ')
```

```python
#optimal solution 
# time complexity: O(n+m) where n is the length of the first list and m is the length of the second list or O(n) if both lists are of equal length

hash_list= [0]*11  #if the array is more than 10 , we can use a dictionary instead of a list to store the frequency of each number in the list = dict()
for num in n: 
    hash_list[num]+=1
for num in m: 
    if num<1 or num >10 : 
        print(0, end=' ')
    else: 
        print(hash_list[num], end=' ')   

```
