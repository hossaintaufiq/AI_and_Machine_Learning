counting number using log

```python
from math import * 

n=int(input("Enter a number: "))
def countDigits(n):
    return floor(log10(n) + 1)

print(countDigits(n))
```

Check Palindrome

```python
n = int(input("Enter a number: "))

num= n; 
def isPalindrome(num): 
  result=0; 
  
  while num>0: 
    last_digit= num%10
    result= result*10+last_digit
    num=num//10 

  return n==result


print(isPalindrome(n))  
```

Armstron number

```python
from math import *
n= int(input("Enter a number: "))

num= n;
def isArmstrong(num):
    result=0; 
    number_count= floor(log10(num)+1)
    while num>0: 
        last_digit= num%10 
        result= result+ last_digit**number_count
        num=num//10 
    return n==result

print(isArmstrong(n)) 
```
