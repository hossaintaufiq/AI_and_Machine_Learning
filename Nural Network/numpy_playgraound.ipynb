```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#creating x values 
x= np.linspace(-10,10,100)\

#mathmetical expression
y = x**(1/12)

# plot 
plt.plot(x,y)

#labels 
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("y= x^1/12")

#show grid
plt.grid()

#display the plot
plt.show()
```

sin(x)

```python
x= np.linspace(-2*np.pi, 2*np.pi, 100 )

y= np.sin(x)

plt.plot(x,y) 

#labels 
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("y= sin(x)")

plt.grid()
plt.show()
```

```python
#understanding the linspace function -> linear space function. Creates numbers that are evenly spaced between a starting value and an ending value. 
x= np.linspace(0,10,200)
print(x)
```

The .arrange function and Damped sine wave 

```python
# x values 
x = np.arange(0,30,.1) 

#expression 
y = np.exp(-.01*x)*np.sin(x) 

#graph 
plt.plot(x,y)

# labels
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("y= e^(-.01x) * sin(x)")

plt.grid()
plt.show()
```

Graph: Two waves Multiplied Together 

y = sin(x)cos(x)

```python
x = np.arange(-10, 10 , .01) 
y = np.sin(x)* np.cos(x)

plt.plot(x,y)

plt.xlabel("x values")
plt.ylabel("y values")
plt.title('y= sin(x)*cos(x)')
plt.grid()
plt.show()
```

Derivatives using numpy and graph using matplotlib 

```python
x= np.arange(-3,3,.01)

#original function 
y = x**3 

#Numerical derivative
dy_dx= np.gradient(y,x)

plt.plot(x,y, label= 'f(x)= x^3')
plt.plot(x,dy_dx, label= "f'(x)= 3x^2")
plt.legend()
plt.grid()
plt.show()

print(dy_dx)
```

```python
import numpy as np 
import matplotlib.pyplot as plt

#main function 
x= np.arange(-5,5,.02)
y = x**3+ 2 * x +6 

# derrivative
dy_dx= np.gradient(y,x)

#labels 
plt.plot(x,y, label= 'f(x)= x^3+ 2 * x +6')
plt.plot(x,dy_dx, label= "f'(x)= 3x^2+ 2")
plt.title("f(x) and f'(x)")
plt.legend()
plt.grid()
plt.show()
```
